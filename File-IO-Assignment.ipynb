```python
'''2. Write a Python program to read first n lines of a file.'''

n = int(input("Enter number of lines to read: "))

with open("sample.txt", "r") as file:
    for i in range(n):
        line = file.readline()

        if not line:
            break

        print(line, end="")
```

```python
'''4. Write a Python program to read last n lines of a file.'''

n = int(input("Enter number of lines to read: "))

with open("sample.txt", "r") as file:
    lines = file.readlines()

# Get last n lines
last_lines = lines[-n:]

print("Last", n, "lines:\n")
for line in last_lines:
    print(line, end="")
```

```python
'''7. Write a python program to find the longest words.'''

with open("sample.txt", "r") as file:
    words = file.read().split()

longest_word = max(words, key=len)

print("Longest word:", longest_word)
```

```python
'''9. Write a Python program to count the frequency of words in a file.'''

with open("sample.txt", "r") as file:
    text = file.read().lower()

words = text.split()
word_count = {}

for word in words:
    if word in word_count:
        word_count[word] += 1
    else:
        word_count[word] = 1

for word, count in word_count.items():
    print(word, ":", count)
```

```python
'''11. Write a Python program to write a list to a file.'''

data = ["Apple", "Banana", "Mango", "Orange"]

with open("sample.txt", "w") as file:
    for item in data:
        file.write(item + "\n")

print("List written to file successfully.")
```

```python
'''12. Write a Python program to copy the contents of a file to another file .'''

with open("sample.txt", "r") as source, open("destination.txt", "w") as dest:
    for line in source:
        dest.write(line)

print("File copied successfully.")
```

```python
'''13. Write a Python program to read a random line from a file.'''

import random

with open("sample.txt", "r") as file:
    lines = file.readlines()

random_line = random.choice(lines)

print("Random line:")
print(random_line)
```

```python
'''15.Write a Python program that takes a text file as input and returns the number of words of a given text file.'''

file_name = input("Enter file name: ")

with open(file_name, "r") as file:
    text = file.read()

# Split text into words
words = text.split()

# Count words
word_count = len(words)

print("Total number of words:", word_count)
```
